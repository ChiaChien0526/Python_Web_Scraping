```python
# 使用 requests 工具
import requests as req
from bs4 import BeautifulSoup as bs

# 匯入 regex 、 os套件
import re
import os
```

```python
# Project Gutenberg 中文書網址
url = "https://www.gutenberg.org/browse/languages/zh"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

# 用 requests 的 get 方法把網頁抓下來
res = req.get(url, headers=headers, timeout=10)

# 指定 lxml 作為解析器
soup = bs(res.text, "html.parser") 

# 建立 list 來放置列表資訊
list_posts = []
list_posts2 = []
```

```python
# 清空放置列表資訊的變數
list_posts.clear()

# 取得中文書的文字與超連結
for a in soup.select('li.pgdbetext > a[href]'):
    text = a.get_text()
    if re.search(r'[\u4E00-\u9FFF]', text):
        print(text)
        print(a['href'])

        # 加入列表資訊
        list_posts.append({
            'title': text,
            'link': 'https://www.gutenberg.org' + a['href']
        })
```

```python
# 清空放置列表資訊的變數
list_posts2.clear()

headers = {
    "User-Agent": "Mozilla/5.0"
}

# 取得中文書網頁中的內文連結
for index, obj in enumerate(list_posts):
    res_ = req.get(obj['link'],  headers=headers)
    soup_ = bs(res_.text, "html.parser")

    for a in soup_.select('td.unpadded.icon_save > a[type="text/plain; charset=us-ascii"]'):
        # Plain Text UTF-8 的標題&連結
        title_tag = soup_.select_one('h1#book_title')
        book_title = title_tag.get_text(strip=True) if title_tag else f"book_{index}"
        match = re.search(r'[\u4e00-\u9fff]{2,}', book_title)
        book_title_clean = match.group(0) if match else book_title
        print(a['href'])

        # 加入列表資訊
        list_posts2.append({
            'title': book_title_clean,
            'link': 'https://www.gutenberg.org' + a['href'] 
        })
```

```python
# 首頁網址
prefix = 'https://www.gutenberg.org'

# 儲存目標資料夾
output_folder = "project_gutenberg"
os.makedirs(output_folder, exist_ok=True)


# 保留：中文字元、中文標點（全形）、換行符號
def extract_clean_chinese(text):
    result = []
    for char in text:
        if re.match(r'[\u4e00-\u9fff]', char):
            result.append(char)
        elif char in '，。？！；：「」『』、．《》（）—':
            result.append(char)
        elif char == '\n':
            result.append(char)
    return ''.join(result)

# 取得中文書中的內容
for index, obj in enumerate(list_posts2):
    full_url = prefix + obj['link'] if obj['link'].startswith('/') else obj['link']
    res_ = req.get(full_url, headers=headers)
    res_.encoding = 'utf-8'

    content = res_.text

    # 保留中文內容
    chinese_text = extract_clean_chinese(content)

    # 用書名做檔名
    title = obj.get('title', f"book_{index}")
    # 移除非法檔名字元，空白轉底線
    clean_title = re.sub(r'[\\/:*?"<>|]', '', title).strip().replace(' ', '_')
    file_path = os.path.join(output_folder, f"{clean_title}.txt")

    
    # 儲存為文字檔
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(chinese_text + '\n\n')

    print(f"已儲存：{file_path}")
```
